```python
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    timeout=10
)
response.raise_for_status()

posts = response.json()

for index, post in enumerate(posts[:5], start=1):
    print(f"Пост {index}")
    print("Заголовок:", post["title"])
    print("Тело:", post["body"])
    print()
```
