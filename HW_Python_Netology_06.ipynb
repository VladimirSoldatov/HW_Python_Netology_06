```python
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    timeout=10
)
response.raise_for_status()

posts = response.json()

for index, post in enumerate(posts[:5], start=1):
    print(f"Пост {index}")
    print("Заголовок:", post["title"])
    print("Тело:", post["body"])
    print()
```

```python
import time
from getpass import getpass

api_key = getpass("Введите API-key: ")
city_name = input("Введите имя города")
response_city = requests.get(
    f"http://api.openweathermap.org/geo/1.0/direct?q={city_name}&limit=5&appid={api_key}",
    timeout=10
)
response_city.raise_for_status()
city_data = response_city.json()

for city in city_data:
    russian_name = city.get("local_names", {}).get("ru")
    if russian_name == city_name:
        print("Широта:", city["lat"])
        print("Долгота:", city["lon"])
        print("Страна:", city["country"])
        print("Регион:", city["state"])
        time.sleep(5)  #Для не получения ошибки 401
        url = "https://api.openweathermap.org/data/2.5/weather"
        params = {
            "lat": city["lat"],
            "lon": city["lon"],
            "appid": api_key,
            "units": "metric",
            "lang": "ru",
        }
        
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        weather = response.json()
        
        print("Город:", weather["name"])
        print("Температура:", weather["main"]["temp"], "°C")
        print()
```
